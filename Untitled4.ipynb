```python
# TASK 2: Phishing Awareness Training - Interactive Quiz

questions = [
    {
        "q": "1. You receive an email asking you to urgently verify your bank account using a link. What should you do?",
        "options": ["A. Click the link immediately", "B. Verify the sender and website before taking action", "C. Share your OTP", "D. Forward it to friends"],
        "answer": "B"
    },
    {
        "q": "2. Which is a common sign of a phishing email?",
        "options": ["A. Unexpected urgency", "B. Official contact information", "C. Expected message from a known person", "D. Normal spelling"],
        "answer": "A"
    },
    {
        "q": "3. What is social engineering?",
        "options": ["A. Designing websites", "B. Manipulating people to reveal information", "C. Installing software", "D. Encrypting files"],
        "answer": "B"
    },
    {
        "q": "4. What should you check before clicking a suspicious link?",
        "options": ["A. Link/URL address", "B. Screen brightness", "C. File size", "D. Wallpaper"],
        "answer": "A"
    },
    {
        "q": "5. What is the safest action if you suspect phishing?",
        "options": ["A. Reply to the email", "B. Click the link to investigate", "C. Report it and delete it", "D. Share your password"],
        "answer": "C"
    }
]

print("=" * 60)
print("        PHISHING AWARENESS TRAINING")
print("=" * 60)
print("\nLearn to identify phishing and stay safe online!\n")

score = 0

for item in questions:
    print(item["q"])
    for option in item["options"]:
        print(option)

    answer = input("Your answer: ").strip().upper()

    if answer == item["answer"]:
        print("✓ Correct!\n")
        score += 1
    else:
        print(f"✗ Incorrect. Correct answer: {item['answer']}\n")

print("=" * 60)
print(f"Quiz Completed! Score: {score}/{len(questions)}")

if score == 5:
    print("Excellent! You have strong phishing awareness.")
elif score >= 3:
    print("Good job! Keep improving your cybersecurity awareness.")
else:
    print("Keep learning! Review common phishing warning signs.")

print("=" * 60)
```
